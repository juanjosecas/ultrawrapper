# 12 - Cinemática básica a partir de tracks

Calcula desplazamiento y velocidad aproximada desde un DataFrame con track_id.


```python
import pandas as pd
```

```python
tracks = pd.DataFrame({
    "frame": [0, 1, 2, 3, 0, 1, 2, 3],
    "track_id": [1, 1, 1, 1, 2, 2, 2, 2],
    "timestamp": [0.0, 0.1, 0.2, 0.3, 0.0, 0.1, 0.2, 0.3],
    "xmin": [10, 14, 18, 22, 80, 78, 76, 74],
    "ymin": [20, 22, 24, 26, 40, 43, 46, 49],
    "xmax": [30, 34, 38, 42, 100, 98, 96, 94],
    "ymax": [60, 62, 64, 66, 80, 83, 86, 89],
})

tracks["center_x"] = (tracks["xmin"] + tracks["xmax"]) / 2
tracks["center_y"] = (tracks["ymin"] + tracks["ymax"]) / 2
tracks
```

## Desplazamiento y velocidad


```python
tracks["dx"] = tracks.groupby("track_id")["center_x"].diff()
tracks["dy"] = tracks.groupby("track_id")["center_y"].diff()
tracks["dt"] = tracks.groupby("track_id")["timestamp"].diff()
tracks["distance_px"] = (tracks["dx"] ** 2 + tracks["dy"] ** 2) ** 0.5
tracks["speed_px_s"] = tracks["distance_px"] / tracks["dt"]
tracks
```

## Resumen por track


```python
kinematics = (
    tracks.groupby("track_id", as_index=False)
          .agg(
              mean_speed_px_s=("speed_px_s", "mean"),
              max_speed_px_s=("speed_px_s", "max"),
              total_distance_px=("distance_px", "sum"),
          )
)
kinematics
```

Para convertir px/s a unidades físicas necesitás una calibración espacial, por ejemplo mm/px.
