# 12 - Tracking kinematics
Convert tracked detections into per-track movement features.

```python
import pandas as pd
from vision.yolo.ml import track_features

TRACKS = "outputs/tracked_predictions.parquet"
FPS = 30.0
df = pd.read_parquet(TRACKS)
features = track_features(df, fps=FPS)
features.head()
```
