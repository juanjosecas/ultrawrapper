# 13 - Explainability
Detection-density and model-agnostic occlusion sensitivity.

```python
import cv2
import matplotlib.pyplot as plt
from vision.yolo.explain import detection_density_map, occlusion_sensitivity, overlay_heatmap
from vision.yolo.infer import predict_image

MODEL = "yolo11n.pt"
IMAGE = "bus.jpg"
image = cv2.imread(IMAGE)
df = predict_image(MODEL, image, confidence=0.25)
```

```python
density = detection_density_map(df, image.shape, sigma=30)
overlay = overlay_heatmap(image, density)
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
def predictor(frame):
    return predict_image(MODEL, frame, confidence=0.25)

sensitivity = occlusion_sensitivity(image, predictor, patch_size=96, stride=96)
overlay = overlay_heatmap(image, sensitivity)
plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
