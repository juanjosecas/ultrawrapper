# 13 - Explainability and density maps

Density maps summarize where detections occur. Occlusion sensitivity estimates which image regions support a detection by hiding one region at a time.


```python
import cv2
from vision.yolo.infer import predict_image
from vision.yolo.explain import detection_density_map, plot_density_map, occlusion_sensitivity

MODEL = "yolo11n.pt"
IMAGE = "notebooks/bus.jpg"
```

```python
image = cv2.imread(IMAGE)
df = predict_image(MODEL, image, confidence=0.25)
df[["class_name", "confidence", "xmin", "ymin", "xmax", "ymax"]].head()
```

```python
height, width = image.shape[:2]
density = detection_density_map(df, image_width=width, image_height=height, bins=32)
plot_density_map(density, image=image)
```

```python
# Slower: the model is run once per grid cell.
heatmap = occlusion_sensitivity(
    MODEL,
    image,
    target_class="person",
    grid_size=8,
    confidence=0.25,
)
plot_density_map(heatmap, image=image)
```
