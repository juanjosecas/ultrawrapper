# 15 - Video I/O without external ffmpeg commands

These helpers use OpenCV directly from Python. No `ffmpeg` subprocess or command-line call is required. Compressed video still depends on native codecs available to OpenCV.


```python
from vision.yolo.video_io import video_info, read_frames, extract_frames, images_to_video

VIDEO = "notebooks/input/4540398-hd_1080_1920_25fps.mp4"
```

```python
info = video_info(VIDEO)
print(info)
```

```python
# Read one frame every 25 frames.
for frame_index, timestamp, frame in read_frames(VIDEO, step=25, max_frames=5):
    print(frame_index, round(timestamp, 2), frame.shape)
```

```python
saved = extract_frames(
    VIDEO,
    output_dir="notebooks/outputs/extracted_frames",
    every_n_frames=25,
    max_frames=20,
)
print(saved[:5])
```

```python
# Rebuild a short MP4 from the extracted images.
output_video = images_to_video(
    saved,
    output_path="notebooks/outputs/rebuilt.mp4",
    fps=5.0,
    codec="mp4v",
)
print(output_video)
```
