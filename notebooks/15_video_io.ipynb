# 15 - Video I/O without ffmpeg CLI
Read metadata, extract frames and rebuild videos using OpenCV.

```python
from vision.yolo.video_io import extract_frames, images_to_gif, images_to_video, video_info

VIDEO = "input/example.mp4"
info = video_info(VIDEO)
info
```

```python
frames = extract_frames(
    VIDEO,
    output_dir="outputs/extracted_frames",
    every=30,
)
frames[:5]
```

```python
images_to_video(frames, "outputs/rebuilt.mp4", fps=10)
images_to_gif(frames[:30], "outputs/preview.gif", duration_ms=100)
```
