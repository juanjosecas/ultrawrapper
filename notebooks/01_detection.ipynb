# 01 - Detección

Ejemplo mínimo y lineal: una imagen, un DataFrame, filtros simples y gráficos.


```python
import os
import pandas as pd

from vision.yolo.infer import predict_image, predict_images
from vision.yolo.plotting import plot_image_detections, plot_class_distribution

NOTEBOOK_DIR = os.path.abspath("notebooks")
OUTPUT_DIR = os.path.join(NOTEBOOK_DIR, "outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)

MODEL = "yolo11n.pt"
IMAGE = os.path.join(NOTEBOOK_DIR, "bus.jpg")
if not os.path.exists(IMAGE):
    IMAGE = "https://ultralytics.com/images/bus.jpg"
```

```python
df = predict_image(MODEL, IMAGE, confidence=0.25)
df.head()
```

```python
high_conf = df[df["confidence"] >= 0.50].copy()
print("Detecciones:", len(df))
print("Confianza >= 0.50:", len(high_conf))
```

```python
plot_image_detections(
    IMAGE,
    df,
    save_to=os.path.join(OUTPUT_DIR, "01_detections.png"),
)

plot_class_distribution(
    df,
    save_to=os.path.join(OUTPUT_DIR, "01_class_distribution.png"),
)
```

## Batch de imágenes

predict_images devuelve un único DataFrame, con una fila por detección.


```python
images = [IMAGE, IMAGE]
batch_df = predict_images(MODEL, images, confidence=0.25, batch_size=2)
batch_df.head()
```
