# 01 - Detection
Basic detection workflow with a DataFrame output.

```python
from vision.yolo.infer import predict_image
from vision.yolo.plotting import plot_image_detections

MODEL = "yolo11n.pt"
IMAGE = "bus.jpg"
OUTPUT = "outputs/detection.png"
```

```python
df = predict_image(MODEL, IMAGE, confidence=0.25)
df.head()
```

```python
plot_image_detections(IMAGE, df, save_to=OUTPUT)
```
