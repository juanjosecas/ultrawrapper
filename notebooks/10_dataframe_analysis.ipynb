# 10 - Análisis tabular de detecciones

Ejemplo de análisis posterior usando únicamente pandas.


```python
import os
import pandas as pd

from vision.yolo.infer import predict_image

NOTEBOOK_DIR = os.path.abspath("notebooks")
MODEL = "yolo11n.pt"
IMAGE = os.path.join(NOTEBOOK_DIR, "bus.jpg")
if not os.path.exists(IMAGE):
    IMAGE = "https://ultralytics.com/images/bus.jpg"
```

```python
df = predict_image(MODEL, IMAGE, confidence=0.20)
df.head()
```

## Resumen por clase


```python
summary = (
    df.groupby("class_name", as_index=False)
      .agg(
          detections=("class_name", "size"),
          mean_confidence=("confidence", "mean"),
          median_confidence=("confidence", "median"),
      )
      .sort_values("detections", ascending=False)
)
summary
```

## Tamaño de los objetos detectados


```python
df["box_width"] = df["xmax"] - df["xmin"]
df["box_height"] = df["ymax"] - df["ymin"]
df["box_area"] = df["box_width"] * df["box_height"]

df[["class_name", "confidence", "box_area"]].head()
```

```python
large_objects = df[df["box_area"] >= df["box_area"].median()].copy()
large_objects.head()
```
