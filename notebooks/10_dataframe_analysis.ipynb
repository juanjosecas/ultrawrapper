# 10 - DataFrame analysis
Simple analysis of saved inference results.

```python
import pandas as pd
import matplotlib.pyplot as plt

PREDICTIONS = "outputs/predictions.parquet"
df = pd.read_parquet(PREDICTIONS)
df.head()
```

```python
summary = df.groupby("class_name").agg(
    n=("class_name", "size"),
    mean_confidence=("confidence", "mean"),
).sort_values("n", ascending=False)
summary
```

```python
df["confidence"].hist(bins=20)
plt.xlabel("Confidence")
plt.ylabel("Detections")
plt.show()
```
