# 11 - Confidence thresholds
Inspect how the number of retained detections changes with the threshold.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("outputs/predictions.parquet")
```

```python
thresholds = np.arange(0.05, 0.96, 0.05)
counts = [(df["confidence"] >= threshold).sum() for threshold in thresholds]

plt.plot(thresholds, counts, marker="o")
plt.xlabel("Confidence threshold")
plt.ylabel("Retained detections")
plt.show()
```
