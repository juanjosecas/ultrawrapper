# 11 - Sensibilidad al umbral de confianza

Compara cómo cambia el número de detecciones al variar confidence.


```python
import os
import pandas as pd
import matplotlib.pyplot as plt

from vision.yolo.infer import predict_image

NOTEBOOK_DIR = os.path.abspath("notebooks")
OUTPUT_DIR = os.path.join(NOTEBOOK_DIR, "outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)

MODEL = "yolo11n.pt"
IMAGE = os.path.join(NOTEBOOK_DIR, "bus.jpg")
if not os.path.exists(IMAGE):
    IMAGE = "https://ultralytics.com/images/bus.jpg"
```

```python
thresholds = [0.10, 0.20, 0.30, 0.40, 0.50]
rows = []

for threshold in thresholds:
    df = predict_image(MODEL, IMAGE, confidence=threshold)
    rows.append({
        "confidence_threshold": threshold,
        "detections": len(df),
        "mean_confidence": df["confidence"].mean() if len(df) else float("nan"),
    })

results = pd.DataFrame(rows)
results
```

```python
ax = results.plot(
    x="confidence_threshold",
    y="detections",
    marker="o",
    legend=False,
)
ax.set_xlabel("Confidence threshold")
ax.set_ylabel("Number of detections")
ax.set_title("Sensitivity to confidence threshold")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "11_confidence_thresholds.png"), dpi=150, bbox_inches="tight")
plt.show()
```
