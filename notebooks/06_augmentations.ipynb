# 06 - Augmentations

Albumentations con variables explícitas y un ejemplo pequeño.


```python
import os
import numpy as np
import matplotlib.pyplot as plt

from vision.yolo.augmentations import build_augmentation_pipeline, apply_augmentation

NOTEBOOK_DIR = os.path.abspath("notebooks")
OUTPUT_DIR = os.path.join(NOTEBOOK_DIR, "outputs")
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
pipeline = build_augmentation_pipeline(
    preset="medium",
    image_size=640,
)
print(pipeline)
```

```python
image = np.random.randint(0, 255, (640, 640, 3), dtype=np.uint8)
bboxes = [[0.5, 0.5, 0.3, 0.4]]
class_labels = [0]

augmented = apply_augmentation(
    pipeline,
    image,
    bboxes=bboxes,
    class_labels=class_labels,
)

print("Imagen:", augmented["image"].shape)
print("Bounding boxes:", augmented.get("bboxes"))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(image)
axes[0].set_title("Original")
axes[0].axis("off")
axes[1].imshow(augmented["image"])
axes[1].set_title("Augmented")
axes[1].axis("off")
fig.tight_layout()
fig.savefig(os.path.join(OUTPUT_DIR, "06_augmentation_preview.png"), dpi=150, bbox_inches="tight")
plt.show()
```
